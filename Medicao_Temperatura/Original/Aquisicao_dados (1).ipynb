# Aquisição dos dados dos sensores T_int e T_ext

```python
import serial
import time
from datetime import datetime
import csv
```

## Configurações

```python
import numpy as np
```

```python

# -------------------------
# LIGAÇÃO ARDUINO
# -------------------------
arduino = serial.Serial('COM3', 9600, timeout=2)
time.sleep(2)

# -------------------------
# CSV
# -------------------------
file = open("temperaturas.csv", "w", newline="")
writer = csv.writer(file)

writer.writerow(["tempo_s", "T_int", "T_ext"])

start_time = time.time()

print("A iniciar leitura... (1 valor por minuto)")

try:
    while True:
        linha = arduino.readline().decode().strip()

        try:
            if linha and linha != "NAN" and linha != "OK":
                parts = linha.split(",")

                t_int = float(parts[0])
                t_ext = float(parts[2])

                tempo = time.time() - start_time

                print(f"{tempo/60:.1f} min | Int: {t_int:.2f}°C | Ext: {t_ext:.2f}°C")

                # guardar dados
                writer.writerow([tempo, t_int, t_ext])
                file.flush()  # 🔥 garante gravação imediata

        except:
            pass

except KeyboardInterrupt:
    print("\nParado pelo utilizador")

finally: 
    # -------------------------
    # FECHAR TUDO CORRETAMENTE
    # -------------------------
    file.close()
    arduino.close()
    print("Ficheiro guardado com sucesso.")
```

## Funções 

```python
import pandas as pd

df = pd.read_csv("temperaturas.csv")

df["tempo_s"] = df["tempo_s"].round().astype(int)

df.to_csv("temperaturas_corrigido.csv", index=False)
```
